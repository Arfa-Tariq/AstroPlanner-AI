```python
from pyngrok import ngrok, conf
import nest_asyncio, uvicorn
import threading
import requests
import time
import os # Import os for system commands

nest_asyncio.apply()

conf.get_default().auth_token = userdata.get("NGROK_TOKEN")

# Kill any processes using port 8000 more aggressively to ensure a clean start
print("Attempting to kill any process on port 8000...")
os.system("lsof -ti tcp:8000 | xargs kill -9 || true") # Use lsof and kill -9
time.sleep(2) # Give a moment for the port to release after aggressive kill

# Configure and run uvicorn server in a separate thread
config = uvicorn.Config("api:app", host="0.0.0.0", port=8000, log_level="info")
server = uvicorn.Server(config)

# Start the server in a new thread. daemon=True ensures the thread exits when the main program does.
global server_thread
server_thread = threading.Thread(target=server.run, daemon=True)
server_thread.start()

# Give Uvicorn time to start and bind to the port, with a health check
print("Waiting for Uvicorn to start and become responsive...")
max_retries = 10
for i in range(max_retries):
    try:
        response = requests.get("http://localhost:8000/health", timeout=1)
        if response.status_code == 200:
            print("Uvicorn server is up and running!")
            break
    except requests.exceptions.ConnectionError:
        pass
    print(f"Attempt {i+1}/{max_retries}: Uvicorn not ready yet, retrying in 1 second...")
    time.sleep(1)
else:
    print("Uvicorn server did not start in time. Requests might fail.")

# Connect ngrok to the now-running Uvicorn server
try:
    # Disconnect existing ngrok tunnels if any, before connecting a new one
    ngrok.kill()
    public_url_obj = ngrok.connect(8000)
    public_url = public_url_obj.url # Extract the actual URL string
    print(f"Ngrok Tunnel URL: {public_url}")
except Exception as e:
    print(f"Error connecting ngrok: {e}")
    public_url = "http://localhost:8000" # Fallback, though likely to fail if ngrok itself fails
    print("Falling back to localhost for requests. This may not work if ngrok failed.")

# Now perform the requests using the correctly extracted public_url
base_url = public_url

# health check
print("Performing health check...")
try:
    health_resp = requests.get(f"{base_url}/health")
    print(f"Health check status: {health_resp.status_code}")
    print(health_resp.json())
except requests.exceptions.ConnectionError as ce:
    print(f"Connection error during health check: {ce}")
except Exception as e:
    print(f"An unexpected error occurred during health check: {e}")

# chat
print("Sending chat request...")
try:
    resp = requests.post(
        f"{base_url}/chat",
        json={
            "user_name": "andrew",
            "thread_id": "s1",
            "message": "Plan tonight at lat 33.2, lon 32.4, 8-inch, 1000mm, beginner"
        }
    )
    print(f"Chat request status: {resp.status_code}")
    print(resp.json())
except requests.exceptions.ConnectionError as ce:
    print(f"Connection error during chat request: {ce}")
except Exception as e:
    print(f"An unexpected error occurred during chat request: {e}")
```
